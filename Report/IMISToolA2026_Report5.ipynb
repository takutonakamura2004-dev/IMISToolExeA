ResNet18を用いた転移学習により、LaptopとSmartphoneを分類する画像分類器を構築した。

データセットは自分で撮影した画像を使用し、LaptopとSmartphoneの2クラスについて、それぞれ学習用40枚、検証用10枚を準備した。

Fine-tuningモデル(model_ft)の学習結果では Validation Accuracy = 1.000 を達成し、固定特徴抽出器として学習した model_conv でも Validation Accuracy = 0.950 という高い性能が得られた。

実験前は、80枚程度の少ない画像数では十分な精度が得られないと思っていた。しかし実際には非常に高い精度で分類することができ、ImageNetで事前学習されたモデルの有効性を実感した。

また、Fine-tuningモデルの方が model_conv よりも高い精度を示したことから、事前学習済みモデル全体を再学習することで、本実験で扱ったLaptopやSmartphoneの特徴により適応できたと考えられる。

一方で、実際にテスト画像を撮影してみると、撮影角度や背景、光の当たり方によって見え方が大きく変化することが分かった。そのため、より実用的な分類器を作るためには、多様な環境で撮影した画像を学習データとして追加する必要があると考えた。

今回の実験を通して、少量のデータでも転移学習を利用することで高性能な画像分類器を比較的容易に構築できることを学んだ。
